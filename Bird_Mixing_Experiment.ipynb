```python

duration = 14
clip_duration = 6
sr = 22050

total_samples = duration * sr
clip_samples = clip_duration * sr

mixed_clean = np.zeros(total_samples, dtype=np.float32)

# Clip each bird recording to exactly 6 seconds
clip_a = bird_a[:clip_samples]
clip_b = bird_b[:clip_samples]

# Place asbfly at 0–6 seconds
mixed_clean[:len(clip_a)] += clip_a

# Place brnhao at 3–9 seconds
start_b = 3 * sr
mixed_clean[start_b:start_b + len(clip_b)] += clip_b

# Place asbfly again at 8–14 seconds
start_a2 = 8 * sr
mixed_clean[start_a2:start_a2 + len(clip_a)] += clip_a

# Prevent clipping while preserving the relative amplitudes
peak = np.max(np.abs(mixed_clean))
if peak > 0:
    mixed_clean = mixed_clean / peak * 0.9

print("Duration:", len(mixed_clean) / sr, "seconds")
print("asbfly intervals: 0–6 s and 8–14 s")
print("brnhao interval: 3–9 s")

display(Audio(mixed_clean, rate=sr))
```

```python

def audio_stats(name, audio):
    rms = np.sqrt(np.mean(audio ** 2))
    peak = np.max(np.abs(audio))

    print(f"{name}")
    print(f"  RMS loudness: {rms:.6f}")
    print(f"  Peak amplitude: {peak:.6f}")
    print()

audio_stats("asbfly original", clip_a)
audio_stats("brnhao original", clip_b)
audio_stats("mixed recording", mixed_clean)
```

```python

def normalize_rms(audio, target_rms=0.03):
    rms = np.sqrt(np.mean(audio ** 2))

    if rms == 0:
        return audio.copy()

    return audio * (target_rms / rms)


# Use 6-second clips
clip_samples = 6 * sr
clip_a = bird_a[:clip_samples].copy()
clip_b = bird_b[:clip_samples].copy()

# Normalize both clips to the same RMS
clip_a_balanced = normalize_rms(clip_a)
clip_b_balanced = normalize_rms(clip_b)

# Build a 14-second mixture
mixed_balanced = np.zeros(14 * sr, dtype=np.float32)

mixed_balanced[0:len(clip_a_balanced)] += clip_a_balanced

start_b = 3 * sr
mixed_balanced[start_b:start_b + len(clip_b_balanced)] += clip_b_balanced

start_a2 = 8 * sr
mixed_balanced[start_a2:start_a2 + len(clip_a_balanced)] += clip_a_balanced

# Prevent clipping without changing the relative balance
peak = np.max(np.abs(mixed_balanced))
if peak > 0.95:
    mixed_balanced *= 0.95 / peak

print("Duration:", len(mixed_balanced) / sr, "seconds")
audio_stats("Balanced asbfly", clip_a_balanced)
audio_stats("Balanced brnhao", clip_b_balanced)
audio_stats("Balanced mixture", mixed_balanced)

display(Audio(mixed_balanced, rate=sr))
```

```python

# Experiment 2: Add controlled Gaussian background noise

rng = np.random.default_rng(42)

signal_power = np.mean(mixed_balanced ** 2)
snr_db = 10

noise = rng.normal(
    loc=0.0,
    scale=1.0,
    size=len(mixed_balanced)
)

noise_power = np.mean(noise ** 2)

# Scale noise to achieve the target SNR
noise *= np.sqrt(
    signal_power / (10 ** (snr_db / 10) * noise_power)
)

mixed_noisy = mixed_balanced + noise

actual_snr = 10 * np.log10(
    np.mean(mixed_balanced ** 2) / np.mean(noise ** 2)
)

print(f"Target SNR: {snr_db} dB")
print(f"Actual SNR: {actual_snr:.2f} dB")
print(f"Duration: {len(mixed_noisy) / sr:.2f} seconds")

display(Audio(mixed_noisy, rate=sr))
```

```python

import cv2
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

def audio_to_model_image(audio, sample_rate=22050):
    # Match the 5-second chunk length used by the existing pipeline
    chunk_samples = 5 * sample_rate
    audio = audio[:chunk_samples]

    if len(audio) < chunk_samples:
        audio = np.pad(audio, (0, chunk_samples - len(audio)))

    # Generate a Mel spectrogram
    mel = librosa.feature.melspectrogram(
        y=audio,
        sr=sample_rate
    )

    mel_db = librosa.power_to_db(mel, ref=np.max)

    # Render it in a way similar to the existing saved spectrograms
    fig, ax = plt.subplots(figsize=(3, 3))
    librosa.display.specshow(
        mel_db,
        sr=sample_rate,
        x_axis="time",
        y_axis="mel",
        ax=ax
    )
    ax.set_axis_off()

    fig.canvas.draw()
    image = np.asarray(fig.canvas.buffer_rgba())[:, :, :3]
    plt.close(fig)

    # Convert to grayscale, resize, and flatten
    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    resized = cv2.resize(gray, (128, 128))
    return resized.flatten()

features = audio_to_model_image(mixed_balanced, sr)

print("Input feature shape:", features.shape)

```

```python

from pathlib import Path
import numpy as np
import librosa
import soundfile as sf
import pandas as pd

root = Path.cwd()
output_dir = root / "multilabel_dataset"
output_dir.mkdir(exist_ok=True)

sr = 22050
duration = 5
samples = sr * duration

asbfly_dir = root / "binary_analysis" / "asbfly"
brnhao_dir = root / "binary_analysis" / "brnhao"

asbfly_files = sorted(asbfly_dir.glob("*.ogg"))
brnhao_files = sorted(brnhao_dir.glob("*.ogg"))

rng = np.random.default_rng(42)
records = []

def load_clip(path):
    audio, _ = librosa.load(path, sr=sr, mono=True)
    if len(audio) < samples:
        audio = np.pad(audio, (0, samples - len(audio)))
    return audio[:samples]

# Create examples containing both species
for i in range(min(len(asbfly_files), len(brnhao_files))):
    a = load_clip(asbfly_files[i])
    b = load_clip(brnhao_files[i])

    # Mix the two recordings at a random relative level
    b_gain = rng.uniform(0.5, 1.0)
    mixture = a + b_gain * b

    peak = np.max(np.abs(mixture))
    if peak > 0:
        mixture = mixture / peak * 0.95

    filename = f"both_{i:03d}.wav"
    sf.write(output_dir / filename, mixture, sr)

    records.append({
        "filename": filename,
        "asbfly": 1,
        "brnhao": 1
    })

# Create single-species examples as well
for species, files in [
    ("asbfly", asbfly_files),
    ("brnhao", brnhao_files)
]:
    for i, path in enumerate(files):
        audio = load_clip(path)
        filename = f"{species}_{i:03d}.wav"
        sf.write(output_dir / filename, audio, sr)

        records.append({
            "filename": filename,
            "asbfly": int(species == "asbfly"),
            "brnhao": int(species == "brnhao")
        })

labels = pd.DataFrame(records)
labels.to_csv(output_dir / "labels.csv", index=False)

print("Dataset folder:", output_dir)
print("Total examples:", len(labels))
print(labels.groupby(["asbfly", "brnhao"]).size())
print("\nLabel preview:")
display(labels.head())
```

```python

from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.multioutput import MultiOutputClassifier
from sklearn.metrics import classification_report
import librosa
import numpy as np
import pandas as pd

# Load labels
labels = pd.read_csv(output_dir / "labels.csv")

# Extract a fixed-length Mel-spectrogram feature vector
def extract_features(filename):
    audio, _ = librosa.load(output_dir / filename, sr=22050, mono=True)

    mel = librosa.feature.melspectrogram(
        y=audio,
        sr=22050,
        n_fft=2048,
        hop_length=512,
        n_mels=64
    )

    mel_db = librosa.power_to_db(mel, ref=np.max)

    # Average over time to create a fixed-size feature vector
    return mel_db.mean(axis=1)

X = np.array([extract_features(f) for f in labels["filename"]])
Y = labels[["asbfly", "brnhao"]].values

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.3, random_state=42
)

model_multi = MultiOutputClassifier(
    make_pipeline(
        StandardScaler(),
        MLPClassifier(
            hidden_layer_sizes=(32,),
            max_iter=1000,
            random_state=42
        )
    )
)

model_multi.fit(X_train, Y_train)

Y_pred = model_multi.predict(X_test)

print("Multi-label classification report:")
print(classification_report(
    Y_test,
    Y_pred,
    target_names=["asbfly", "brnhao"],
    zero_division=0
))
```

```python

import numpy as np
import pandas as pd
import librosa
import soundfile as sf
from pathlib import Path

train_dir = root / "multilabel_train"
train_dir.mkdir(exist_ok=True)

sr = 22050
duration = 5
n_samples = sr * duration
rng = np.random.default_rng(42)

def load_five_seconds(path):
    audio, _ = librosa.load(path, sr=sr, mono=True)

    if len(audio) < n_samples:
        audio = np.pad(audio, (0, n_samples - len(audio)))

    return audio[:n_samples]

def save_audio(audio, filename):
    peak = np.max(np.abs(audio))
    if peak > 0.99:
        audio = audio * (0.99 / peak)

    sf.write(train_dir / filename, audio, sr)

rows = []

# Single-species examples from training recordings only
for species, paths in [
    ("asbfly", asbfly_train),
    ("brnhao", brnhao_train)
]:
    for i, path in enumerate(paths):
        audio = load_five_seconds(path)
        filename = f"{species}_{i:02d}.wav"
        save_audio(audio, filename)

        rows.append({
            "filename": filename,
            "asbfly": int(species == "asbfly"),
            "brnhao": int(species == "brnhao")
        })

# Mixed examples using training recordings only
for i in range(min(len(asbfly_train), len(brnhao_train))):
    a = load_five_seconds(asbfly_train[i])
    b = load_five_seconds(brnhao_train[i])

    # Randomly vary the relative volume of the two sources
    b *= rng.uniform(0.5, 1.0)
    mixture = a + b

    filename = f"both_{i:02d}.wav"
    save_audio(mixture, filename)

    rows.append({
        "filename": filename,
        "asbfly": 1,
        "brnhao": 1
    })

train_labels = pd.DataFrame(rows)
train_labels.to_csv(train_dir / "labels.csv", index=False)

print("Training examples:", len(train_labels))
display(train_labels.groupby(["asbfly", "brnhao"]).size())
display(train_labels.head())
```

```python

from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.multioutput import MultiOutputClassifier
from sklearn.pipeline import make_pipeline
from sklearn.metrics import classification_report
import numpy as np
import librosa

def extract_mel_features(path):
    audio, _ = librosa.load(path, sr=22050, mono=True)

    mel = librosa.feature.melspectrogram(
        y=audio,
        sr=22050,
        n_fft=2048,
        hop_length=512,
        n_mels=64
    )

    mel_db = librosa.power_to_db(mel, ref=np.max)

    # Resize the time-frequency representation to a fixed shape
    from PIL import Image
    image = Image.fromarray(
        np.uint8(np.clip((mel_db + 80) / 80 * 255, 0, 255))
    )
    image = image.resize((32, 32))

    return np.asarray(image, dtype=np.float32).flatten()

X_train = np.array([
    extract_mel_features(train_dir / filename)
    for filename in train_labels["filename"]
])

Y_train = train_labels[["asbfly", "brnhao"]].values

model_multi = MultiOutputClassifier(
    make_pipeline(
        StandardScaler(),
        MLPClassifier(
            hidden_layer_sizes=(32,),
            max_iter=1000,
            random_state=42
        )
    )
)

model_multi.fit(X_train, Y_train)

print("Training complete!")
print("Feature shape:", X_train.shape)
print("Label shape:", Y_train.shape)
```

```python

# Build a test set from recordings held out before training

test_rows = []

for species, paths in [
    ("asbfly", asbfly_test),
    ("brnhao", brnhao_test)
]:
    for i, path in enumerate(paths):
        filename = f"{species}_test_{i}.wav"
        audio = load_five_seconds(path)
        sf.write(root / filename, audio, sr)

        test_rows.append({
            "path": root / filename,
            "asbfly": int(species == "asbfly"),
            "brnhao": int(species == "brnhao")
        })

X_test = np.array([
    extract_mel_features(row["path"])
    for row in test_rows
])

Y_test = np.array([
    [row["asbfly"], row["brnhao"]]
    for row in test_rows
])

Y_pred = model_multi.predict(X_test)

print("Actual labels [asbfly, brnhao]:")
print(Y_test)

print("\nPredicted labels [asbfly, brnhao]:")
print(Y_pred)

print("\nPer-species report:")
print(classification_report(
    Y_test,
    Y_pred,
    target_names=["asbfly", "brnhao"],
    zero_division=0
))
```
